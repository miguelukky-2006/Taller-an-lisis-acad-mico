
# EJERCICIO 3. DISTRIBUCIÓN DE LOS RESULTADOS ACADÉMICOS

## Objetivo
Identificar cómo se distribuyen los estudiantes según su resultado académico final.

## Desarrollo
Se utiliza Pandas para contar los estudiantes de cada categoría de la variable Target. Posteriormente, se calculan los porcentajes correspondientes y se obtiene la edad promedio de cada grupo.

Las categorías analizadas son:
- Graduate: estudiantes graduados.
- Dropout: estudiantes que abandonaron sus estudios.
- Enrolled: estudiantes que continúan matriculados.

## Análisis
Se busca determinar:
- Cuál es el resultado académico más frecuente.
- Qué porcentaje de estudiantes abandonó sus estudios.
- Qué diferencias existen entre las edades promedio de los grupos.

## Resultado esperado
Comprender la distribución de los resultados académicos y reconocer las diferencias generales entre los grupos.


```python

# EJERCICIO 4: COMPARACIÓN DEL RENDIMIENTO ACADÉMICO POR SEMESTRE

# 1. Identificar las columnas
col_nota_1 = 'Curricular units 1st sem (grade)'
col_nota_2 = 'Curricular units 2nd sem (grade)'
col_aprobadas_1 = 'Curricular units 1st sem (approved)'
col_aprobadas_2 = 'Curricular units 2nd sem (approved)'

# 2. Convertir las calificaciones en arreglos NumPy
nota_1 = df[col_nota_1].to_numpy()
nota_2 = df[col_nota_2].to_numpy()

# 3. Calcular promedios generales
print("PROMEDIO GENERAL DE CALIFICACIONES")

print("Primer semestre:", round(np.mean(nota_1), 2))
print("Segundo semestre:", round(np.mean(nota_2), 2))

# 4. Comparar promedios por resultado académico
comparacion = df.groupby('Target')[
    [col_nota_1, col_nota_2, col_aprobadas_1, col_aprobadas_2]
].mean()

print("\nCOMPARACIÓN POR RESULTADO ACADÉMICO:")
display(comparacion.round(2))

# 5. Calcular diferencia individual entre semestres
df['Diferencia_notas'] = df[col_nota_2] - df[col_nota_1]

diferencias = df['Diferencia_notas'].to_numpy()

# 6. Contar estudiantes según su evolución
print("\nEVOLUCIÓN DE LAS CALIFICACIONES")

print("Mejoraron:", np.sum(diferencias > 0))
print("Disminuyeron:", np.sum(diferencias < 0))
print("Mantuvieron:", np.sum(diferencias == 0))
```

También se calcula la diferencia entre las calificaciones del segundo semestre y las del primero para identificar cuántos estudiantes mejoraron, disminuyeron o mantuvieron sus resultados.

Se busca identificar:

```python

# EJERCICIO 5: ESTADO DE MATRÍCULA Y BECAS

# 1. Convertir columnas en arreglos NumPy
matricula = df['Tuition fees up to date'].to_numpy()
beca = df['Scholarship holder'].to_numpy()

# 2. Contar estudiantes según el estado de matrícula
al_dia = np.sum(matricula == 1)
pendiente = np.sum(matricula == 0)

print("ESTADO DE MATRÍCULA")
print("Matrícula al día:", al_dia)
print("Matrícula pendiente:", pendiente)

# 3. Calcular porcentajes
print("\nPORCENTAJES DE MATRÍCULA")
print("Al día:", round(al_dia / len(df) * 100, 2), "%")
print("Pendiente:", round(pendiente / len(df) * 100, 2), "%")

# 4. Contar estudiantes con beca y sin beca
print("\nESTADO DE BECAS")
print("Estudiantes con beca:", np.sum(beca == 1))
print("Estudiantes sin beca:", np.sum(beca == 0))

# 5. Tabla cruzada: matrícula frente a resultado académico
print("\nMATRÍCULA FRENTE A RESULTADO ACADÉMICO")
display(pd.crosstab(
    df['Tuition fees up to date'],
    df['Target']
))

# 6. Tabla cruzada: becas frente a resultado académico
print("\nBECAS FRENTE A RESULTADO ACADÉMICO")
display(pd.crosstab(
    df['Scholarship holder'],
    df['Target']
))
```

## Objetivo

Se utiliza Matplotlib para construir cuatro visualizaciones:
1. Gráfico de barras: cantidad de estudiantes por resultado académico.

Comunicar visualmente los principales comportamientos de la población estudiantil y facilitar la identificación de tendencias.


```python

# EJERCICIO 7: FACTORES ASOCIADOS AL RESULTADO ACADÉMICO

# 1. Seleccionar las variables para el análisis
variables = [
    'Age at enrollment',
    'Admission grade',
    'Curricular units 1st sem (grade)',
    'Curricular units 1st sem (approved)',
    'Curricular units 2nd sem (grade)',
    'Curricular units 2nd sem (approved)'
]

# 2. Calcular la matriz de correlación con Pandas
matriz_corr = df[variables].corr()

print("MATRIZ DE CORRELACIÓN:")
display(matriz_corr.round(2))

# 3. Calcular correlación entre las notas con NumPy
nota_1 = df['Curricular units 1st sem (grade)'].to_numpy()
nota_2 = df['Curricular units 2nd sem (grade)'].to_numpy()

corr_numpy = np.corrcoef(nota_1, nota_2)[0, 1]

print("\nCORRELACIÓN ENTRE LAS NOTAS DE AMBOS SEMESTRES:")
print(round(corr_numpy, 4))

# 4. Crear el mapa de calor de correlaciones
plt.figure(figsize=(9, 7))

plt.imshow(matriz_corr, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(label='Coeficiente de correlación')

plt.xticks(
    range(len(variables)),
    variables,
    rotation=75,
    ha='right'
)

plt.yticks(range(len(variables)), variables)

for i in range(len(variables)):
    for j in range(len(variables)):
        plt.text(
            j, i,
            f'{matriz_corr.iloc[i, j]:.2f}',
            ha='center',
            va='center',
            fontsize=8
        )

plt.title('Matriz de correlación de variables académicas')
plt.tight_layout()
plt.show()

# 5. Crear resumen por resultado académico
resumen_final = df.groupby('Target').agg(
    estudiantes=('Target', 'size'),
    edad_promedio=('Age at enrollment', 'mean'),
    nota_semestre_1=('Curricular units 1st sem (grade)', 'mean'),
    nota_semestre_2=('Curricular units 2nd sem (grade)', 'mean'),
    aprobadas_semestre_1=('Curricular units 1st sem (approved)', 'mean'),
    aprobadas_semestre_2=('Curricular units 2nd sem (approved)', 'mean')
)

print("\nRESUMEN FINAL POR RESULTADO ACADÉMICO:")
display(resumen_final.round(2))
```
